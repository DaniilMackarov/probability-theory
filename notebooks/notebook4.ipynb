## Парадокс Монти Холла
> Перед игроком три закрытые двери. За одной находится автомобиль, за двумя другими — козы. Игрок выбирает одну дверь. Ведущий, знающий расположение автомобиля, открывает одну из двух остальных дверей, за которой находится коза, и предлагает изменить выбор на оставшуюся закрытую дверь. Какая стратегия выгоднее: сохранить выбор или сменить дверь?

---

**Условия:** автомобиль с равной вероятностью находится за каждой дверью; ведущий всегда открывает невыбранную дверь с козой и всегда предлагает сменить выбор. Если он может открыть две двери, то выбирает одну из них случайно.

### Теоретическое решение

- Если первоначальный выбор правильный (вероятность $1/3$), сохранение выбора приносит выигрыш, а смена двери — проигрыш.
- Если первоначальный выбор неправильный (вероятность $2/3$), ведущий открывает единственную другую дверь с козой. За оставшейся закрытой дверью обязательно находится автомобиль, поэтому смена двери приносит выигрыш.

$$P(\text{выигрыш без смены})=\frac{1}{3},\qquad
P(\text{выигрыш со сменой})=\frac{2}{3}.$$

Выгоднее сменить дверь

### Моделирование

```python
from random import Random


def simulate_monty_hall(n, seed=42):
    rng = Random(seed)
    doors = (0, 1, 2)
    wins_stay = 0
    wins_switch = 0

    for _ in range(n):
        car = rng.choice(doors)
        chosen = rng.choice(doors)

        # Ведущий открывает невыбранную дверь, за которой нет автомобиля.
        available = [d for d in doors if d != chosen and d != car]
        opened = rng.choice(available)

        # При смене выбираем другую закрытую дверь.
        switched = next(d for d in doors if d != chosen and d != opened)

        wins_stay += chosen == car
        wins_switch += switched == car

    return wins_stay, wins_switch

n = 100_000
wins_stay, wins_switch = simulate_monty_hall(n)
p_stay = wins_stay / n
p_switch = wins_switch / n

print(f"Количество испытаний: {n}")
print(f"Без смены: {wins_stay} выигрышей; вероятность ≈ {p_stay:.4f}")
print(f"Со сменой: {wins_switch} выигрышей; вероятность ≈ {p_switch:.4f}")
```
