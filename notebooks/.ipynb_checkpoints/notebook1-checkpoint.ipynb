## Тема 1 - Моделирование геометрической вероятности
---

### Задача
> Смоделируйте вычисление геометрической вероятности (на примере из лекции про метание дротиков в игровой круг, площадь которого в два раза меньше >площади стены). Показать теоретический расчет и графическую иллюстрацию для рассматриваемого примера.


* У круга площадь в k раз меньше площади стены: $$S_{\text{круга}} = \frac{1}{k}S_{\text{стены}}$$
Благоприятный исход - это попадание в мишень: $$P(A) =\frac{S_{\text{круга}}}{S_{\text{стены}}}$$
Значит: $$P(A) =\frac{\frac{S_{\text{стены}}}{k}}{S_{\text{стены}}}= \frac{1}{k}$$
В среднем доля попадания дротиков при k = 2 равна: $$\boxed{P(A) = \frac{1}{2} = 0{,}5}$$


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle

x0, y0, R = 6, 6, 5
size = 500
side = np.sqrt(2 * np.pi * R ** 2)
fig, ax = plt.subplots(figsize=(5, 5))
ax.add_patch(Rectangle((0, 0), side, side))
ax.add_patch(Circle((x0, y0), R, facecolor='r'))

X = np.random.uniform(0, side, size)
Y = np.random.uniform(0, side, size)
plt.scatter(X, Y, marker='*', c='black')

inside = len([d for d in (X-x0)**2 + (Y - y0)**2 if d <= R**2])
prob = inside / size

plt.show()
print(f"В круг попало {inside} из {size} дротиков. Доля попавших составила {prob:.4f}.")
```
