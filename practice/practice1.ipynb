## Практика 4. Полная вероятность
---

### Задача
> Решить классически и смоделировать решение задачи. В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.
---

### Обозначения

После перекладывания 3 шаров состав второй урны изменится. Возможны 4 гипотезы:

| Гипотеза | Переложено белых | Белых в урне |
|:--------:|:----------------:|:--------------:|
| $H_0$ | 0 | $6$ | 
| $H_1$ | 1 | $7$ | 
| $H_2$ | 2 | $8$ | 
| $H_3$ | 3 | $9$ | 

---

### Вероятности

Общее число способов выбрать 3 шара из 20:

$$
C_{20}^{3} = \frac{20 \cdot 19 \cdot 18}{1 \cdot 2 \cdot 3} = 1140
$$

Значит вероятность i-го события:

$$
P(H_i) = \frac{C_{15}^{i} \cdot C_{5}^{3-i}}{C_{20}^{3}}
$$

| Гипотеза | $P(H_i)$ |
|:--------:|:--------:|
| $H_0$ | $\dfrac{10}{1140}$ |
| $H_1$ | $\dfrac{150}{1140}$ |
| $H_2$ | $\dfrac{525}{1140}$ |
| $H_3$ | $\dfrac{455}{1140}$ |

---

### Условные вероятности

Если верна гипотеза $H_i$, то во второй стало $(6+i)$ белых из 10 шаров:

$$
P(\text{белый} \mid H_i) = \frac{6+i}{10}
$$

| Гипотеза| $P(\text{белый} \mid H_i)$ |
|:--------:|:--------------------------:|
| $H_0$ | $6/10$ |
| $H_1$ | $7/10$ |
| $H_2$ | $8/10$ |
| $H_3$ | $9/10$ |

---

### Формула полной вероятности

$$
P(\text{белый}) = \sum_{i=0}^{3} P(H_i) \cdot P(\text{белый} \mid H_i)
$$


$$
P = \frac{10}{1140} \cdot \frac{6}{10}
  + \frac{150}{1140} \cdot \frac{7}{10}
  + \frac{525}{1140} \cdot \frac{8}{10}
  + \frac{455}{1140} \cdot \frac{9}{10}
    = \frac{9405}{11400} = \frac{627}{760} \approx 0{,}825
$$

---

```python
import random
import math
from collections import Counter

URNA1_WHITE = 15
URNA1_BLACK = 5
URNA2_WHITE = 6
URNA2_BLACK = 1

 # число экспериментов
N = 1_000_000 

def experiment():
    urna1 = ['W'] * 15 + ['B'] * 5
    
    moved = random.sample(urna1, 3)
    
    urna2 = ['W'] * 6 + ['B'] * 1
    urna2.extend(moved)
    
    final = random.choice(urna2)
    return final

random.seed(42)
results = [experiment() for _ in range(N)]

white_count = results.count('W')
probability = white_count / N

print(f"Число экспериментов: {N:,}")
print(f"Белых шаров вынуто: {white_count:,}")
print(f"Частота белого: {probability:.6f}")
print()

```

### Задача 2.9
> В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того, что из 9 пакетов акций: 1) не будут проданы по первоначально заявленной цене 5 пакетов; 2) будут проданы по этой цене: а) менее 2 пакетов; б) не более 2 пакетов; в) хотя бы 2 пакета; г) найти наиболее вероятное число проданных пакетов.

---

Пусть $X$ — число пакетов, проданных по первоначальной цене. 

Применяем формулу Бернулли


```python
from scipy.stats import binom

n = 9
p = 0.2
```

- **1)** Пусть событие A — не проданы ровно 5 пакетов. Это означает, что проданы ровно 4 пакета: $X=4$.

```python
round(float(binom.pmf(4, n, p)), 4)
```

- **2а)** Пусть событие B — проданы менее 2 пакетов: $X<2$, то есть $X\leq1$.

```python
round(float(binom.cdf(1, n, p)), 4)
```

- **2б)** Пусть событие C — проданы не более 2 пакетов: $X\leq2$.

```python
round(float(binom.cdf(2, n, p)), 4)
```

- **2в)** Пусть событие D — проданы хотя бы 2 пакета: $X\geq2$

```python
round(float(binom.sf(1, n, p)), 4)
```

- **2г)** Наиболее вероятное число пакетов удовлетворяет условию $(n+1)p-1\leq m_0\leq(n+1)p$. Здесь $1\leq m_0\leq2$, поэтому наиболее вероятны **1 и 2 пакета**.

```python
modes = [1, 2]
tuple(round(float(binom.pmf(k, n, p)), 4) for k in modes)
```

Вероятность того, что число проданных пакетов равно одному из наиболее вероятных значений:

```python
round(float(binom.pmf(1, n, p) + binom.pmf(2, n, p)), 4)
```

Здесь $n=9$ мало и $np(1-p)=1{,}44$. Используем точные вероятности

## Задача 2.10

> Завод отправил на базу 10 000 стандартных изделий. Средняя доля изделий, повреждаемых при транспортировке, составляет 0,02%. Найти вероятность того, что из 10 000 изделий: 1) будет повреждено: а) 3; б) по крайней мере 3; 2) не будет повреждено: а) 9997; б) хотя бы 9997.

---

Пусть $D$ — число повреждённых изделий. 

Применяем приближение Пуассона

Пусть **первое число — точная вероятность, второе — приближение Пуассона**.

```python
from scipy.stats import binom, poisson

n = 10000
p = 0.0002
lam = n * p
```

- **1а)** Пусть событие A — повреждены ровно 3 изделия: $D=3$.

```python
(round(float(binom.pmf(3, n, p)), 4),
 round(float(poisson.pmf(3, lam)), 4))
```

- **1б)** Пусть событие B — повреждены по крайней мере 3 изделия: $D\geq3$.

```python
(round(float(binom.sf(2, n, p)), 4),
 round(float(poisson.sf(2, lam)), 4))
```

- **2а)** Пусть событие C — не повреждены ровно 9997 изделий. Это равносильно событию $D=3$.

```python
(round(float(binom.pmf(3, n, p)), 4),
 round(float(poisson.pmf(3, lam)), 4))
```

- **2б)** Пусть событие E — не повреждены хотя бы 9997 изделий. Тогда $10000-D\geq9997$, то есть $D\leq3$.

```python
(round(float(binom.cdf(3, n, p)), 4),
 round(float(poisson.cdf(3, lam)), 4))
```

Приближение Пуассона подходит($n$ велико, а $p$ мало). Приближение Муавра—Лапласа не применяем

---


Пусть $X$ — число предприятий с нарушениями.

```python
from math import sqrt
from scipy.stats import binom, norm

n = 1000
p = 0.5
mu = n * p
sigma = sqrt(n * p * (1 - p))
```

- **а)** Пусть событие A — нарушения имеют ровно 480 предприятий: $X=480$.

```python
(round(float(binom.pmf(480, n, p)), 4),
 round(float(norm.pdf((480 - mu) / sigma) / sigma), 4))
```

- **б)** 


```python
m0 = int((n + 1) * p)
m0
```

```python
(round(float(binom.pmf(m0, n, p)), 4),
 round(float(norm.pdf((m0 - mu) / sigma) / sigma), 4))
```

- **в)** Пусть событие B — нарушения имеют не менее 480 предприятий

Применяем интегральную формулу


```python
(round(float(binom.sf(479, n, p)), 4),
 round(float(norm.sf((480 - mu) / sigma)), 4),
 round(float(norm.sf((479.5 - mu) / sigma)), 4))
```

- **г)** Пусть событие C — нарушения имеют от 480 до 520 предприятий включительно

```python
(round(float(binom.cdf(520, n, p) - binom.cdf(479, n, p)), 4),
 round(float(norm.cdf((520 - mu) / sigma) - norm.cdf((480 - mu) / sigma)), 4),
 round(float(norm.cdf((520.5 - mu) / sigma) - norm.cdf((479.5 - mu) / sigma)), 4))
```
