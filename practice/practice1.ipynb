### Задача 1.10
>Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой:
>а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* а) Пусть событие А - получение слова «ТОР».

### Задача 1.16
>За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом;
>б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

### Задача 1.17
>В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе;
>в) в три купе. 

* а) Пусть событие А - пассажиры попали в два купе.

```python
round(sc.comb(9, 2) * sc.comb(8, 7) / sc.comb(36, 7), 5)
```

* б) Пусть событие В - пассажиры попали в семь купе.

```python
round(sc.comb(9, 7) * 4 ** 7 / (sc.comb(36, 7)) , 4)
```

* в) Пусть событие D - пассажиры попали в три купе.

```python
m3 = sc.comb(9, 3) *(6 * sc.comb(4,1)*sc.comb(4,2)*sc.comb(4,4) + 3 * sc.comb(4,1)*sc.comb(4,3)*sc.comb(4,3) + 3 * sc.comb(4,2)*sc.comb(4,2)*sc.comb(4,3))
round(m3 / sc.comb(36, 7) , 4) 
```

### Задача 1.48
>В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а)выпадения 11 очков; б) выигрыша. 

 * в) Пусть событие A - выпало 11 очков

```python
outcomes = list(it.product(range(1 , 7), repeat = 3))
variants_11 = [o for o in outcomes if sum(o) == 11]
round(len(variants_11) / len(outcomes), 4)
```

* в) Пусть событие В - в игре набралось больше 10 очков

```python
variants_win = [o for o in outcomes if sum(o) > 10]
round(len(variants_win) / len(outcomes), 4)
```
