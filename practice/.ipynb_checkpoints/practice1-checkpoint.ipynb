### Задача 1.48
>В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а)выпадения 11 очков; б) выигрыша.
---

 * в) Пусть событие A - выпало 11 очков

```python
outcomes = list(it.product(range(1 , 7), repeat = 3))
variants_11 = [o for o in outcomes if sum(o) == 11]
round(len(variants_11) / len(outcomes), 4)
```

* в) Пусть событие В - в игре набралось больше 10 очков

```python
variants_win = [o for o in outcomes if sum(o) > 10]
round(len(variants_win) / len(outcomes), 4)
```

## Практика 3. Полная вероятность
---

### Задача
> Решить классически и смоделировать решение задачи. В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.
---

### Обозначения

После перекладывания 3 шаров состав второй урны изменится. Возможны 4 гипотезы:

| Гипотеза | Переложено белых | Белых в урне |
|:--------:|:----------------:|:--------------:
| $H_0$ | 0 | $6$ | 
| $H_1$ | 1 | $7$ | 
| $H_2$ | 2 | $8$ | 
| $H_3$ | 3 | $9$ | 

---

### Вероятности

Общее число способов выбрать 3 шара из 20:

$$
C_{20}^{3} = \frac{20 \cdot 19 \cdot 18}{1 \cdot 2 \cdot 3} = 1140
$$

Значит вероятность i-го события:

$$
P(H_i) = \frac{C_{15}^{i} \cdot C_{5}^{3-i}}{C_{20}^{3}}
$$

| Гипотеза | $P(H_i)$ |
|:--------:|:--------:|
| $H_0$ | $\dfrac{10}{1140}$ |
| $H_1$ | $\dfrac{150}{1140}$ |
| $H_2$ | $\dfrac{525}{1140}$ |
| $H_3$ | $\dfrac{455}{1140}$ |

---

### Условные вероятности

Если верна гипотеза $H_i$, то во второй стало $(6+i)$ белых из 10 шаров:

$$
P(\text{белый} \mid H_i) = \frac{6+i}{10}
$$

| Гипотеза| $P(\text{белый} \mid H_i)$ |
|:--------:|:--------------------------:|
| $H_0$ | $6/10$ |
| $H_1$ | $7/10$ |
| $H_2$ | $8/10$ |
| $H_3$ | $9/10$ |

---

### Формула полной вероятности

$$
P(\text{белый}) = \sum_{i=0}^{3} P(H_i) \cdot P(\text{белый} \mid H_i)
$$


$$
P = \frac{10}{1140} \cdot \frac{6}{10}
  + \frac{150}{1140} \cdot \frac{7}{10}
  + \frac{525}{1140} \cdot \frac{8}{10}
  + \frac{455}{1140} \cdot \frac{9}{10}
    = \frac{9405}{11400} = \frac{627}{760} \approx 0{,}825
$$

---

```python
import random
import math
from collections import Counter

URNA1_WHITE = 15
URNA1_BLACK = 5
URNA2_WHITE = 6
URNA2_BLACK = 1

 # число экспериментов
N = 1_000_000 

def experiment():
    urna1 = ['W'] * 15 + ['B'] * 5
    
    moved = random.sample(urna1, 3)
    
    urna2 = ['W'] * 6 + ['B'] * 1
    urna2.extend(moved)
    
    final = random.choice(urna2)
    return final

random.seed(42)
results = [experiment() for _ in range(N)]

white_count = results.count('W')
probability = white_count / N

print(f"Число экспериментов: {N:,}")
print(f"Белых шаров вынуто: {white_count:,}")
print(f"Частота белого: {probability:.6f}")
print()

```
